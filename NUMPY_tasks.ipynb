```python
#task 5
import numpy as np

marks = np.array([35, 67, 89, 45, 90, 22, 70, 58])

passed_marks = marks[marks >= 50]
failed_marks = marks[marks < 50]

print("Passed marks:", passed_marks)
print("Failed marks:", failed_marks)
print("Number of passed:", len(passed_marks))
print("Number of failed:", len(failed_marks))
```

```python
#Python List vs NumPy Array

marks = [50, 90, 80, 30]
import numpy as np

marks = np.array([50, 90, 80, 30])

print(marks)
```

```python
#Creating NumPy Arrays
numbers = np.array([50, 60, 30, 40])

print(numbers)
```

```python
#Two-Dimensional array

marks = np.array([
    [70, 80, 90],
    [60, 75, 85],
    [88, 92, 95]
])

print(marks)
```

```python
#import array properties
data = np.array([
    [10, 20, 30],
    [40, 50, 60]
])
print(data.ndim)
print(data.shape)
print(data.size)
print(data.dtype)
```

```python
#Indexing
numbers = np.array([100,200,300,500,600])
print(numbers[4])
print(numbers[3])

#negative indexing

print(numbers[-5])
```

```python
#Indexing in a 2D Array
marks = np.array([
    [70, 80, 90],
    [60, 75, 85]
])
print(marks[0, 1])
```

```python
#slicing
numbers = np.array([10, 20, 30, 40, 50])

print(numbers[1:4])
```

```python
#Vectorized Operations
numbers = np.array([20,30,90,60])
print(numbers + 15)
print(numbers * 2)
print(numbers - 10)
print(numbers / 5)
```

```python
#Operations Between Two arrays
a = np.array([40, 50, 70])
b = np.array([1, 5, 10])
print(a + b)
print(a - b)
print(a * b)
print(a / b )
```

```python
#Mathematical Functions
numbers = np.array([4, 9, 16, 25])

print(np.sqrt(numbers))
numbers = np.array([2, 3, 4])

print(np.power(numbers, 2))
```

```python
#Absolute
numbers = np.array([-10,20,30])
print(np.abs(numbers))
```

```python
#Round
numbers = np.array([6.2, 4.5, 7.8, 9.3])
print(np.round(numbers))
```

```python
#Aggregation Functions
marks = np.array([10, 20, 30, 40])
print(np.sum(marks))
print(np.max(marks))
print(np.min(marks))
print(np.mean(marks))
print(np.median(marks))
print(np.std(marks))
```

```python
#Conditions in NumPy
marks = np.array([30, 50, 80, 45, 90, 20])

print(marks >= 50)
```

```python
#Filtering Data
marks = np.array([30, 50, 80, 45, 90, 20])

passed = marks[marks >= 50]

print(passed)
```

```python
failed = marks[marks < 50]
print(failed)
```

```python
#Reshape
numbers = np.array([1, 2, 3, 4, 5, 6])
print(numbers)
new_array = numbers.reshape(2, 3)

print(new_array)
```

```python
#Flatten
data = np.array([
    [1, 2, 3],
    [4, 5, 6]
])

print(data.flatten())
```

```python
#useful NumPy Functions
print(np.zeros(5))
print(np.ones(5))

print(np.arange(1, 11))
```
