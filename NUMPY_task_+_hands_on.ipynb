```python
# Task 1 - Create Array and Find Sum, Average, Max, Min

# 1. Create this array
arr = [20, 30, 40, 50, 60]
print("Array:", arr)

# 2. Find Sum
total_sum = sum(arr)
print("Sum:", total_sum)

# 3. Find Average
average = total_sum / len(arr)
print("Average:", average)

# 4. Find Maximum
maximum = max(arr)
print("Maximum:", maximum)

# 5. Find Minimum
minimum = min(arr)
print("Minimum:", minimum)
```

```python
# Task 4 - Generate 10 random marks between 40 and 100
import random
import numpy as np

# 1. Generate 10 random marks between 40 and 100
marks = np.random.randint(40, 101, 10)

print(f"Random Marks: {marks}")

# 2. Find Average
average = sum(marks) / len(marks)
print(f"Average: {average:.2f}")

# 3. Find Highest marks
highest = max(marks)
print(f"Highest Marks: {highest}")

# 4. Find Lowest marks
lowest = min(marks)
print(f"Lowest Marks: {lowest}")
```

```python
# Task 5 - Find Passed and Failed marks
import numpy as np

# 1. Create this marks array
marks = np.array([40, 70, 99, 55, 80, 22, 50, 58])
passing_marks = 50

print(f"All Marks: {marks}")
print(f"Passing Marks: {passing_marks}\n")

# 2. Passed marks (>= 50)
passed_marks = marks[marks >= passing_marks]
print(f"Passed Marks: {passed_marks}")

# 3. Failed marks (< 50)
failed_marks = marks[marks < passing_marks]
print(f"Failed Marks: {failed_marks}")

# 4. Number of passed students
num_passed = len(passed_marks)
print(f"Number of Passed Students: {num_passed}")

# 5. Number of failed students
num_failed = len(failed_marks)
print(f"Number of Failed Students: {num_failed}")
```
